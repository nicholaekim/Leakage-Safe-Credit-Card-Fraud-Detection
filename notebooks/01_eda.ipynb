```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src import config, data
```

```python
df = data.load_raw()
rep = data.integrity_report(df)
rep
```

## Class balance & duplicates

the imbalance is extreme (~0.17% fraud). exact duplicate rows get dropped before splitting - a dupe landing on both sides of the split is leakage.

```python
print('Fraud rate: {:.4f}%'.format(100 * rep['fraud_rate']))
print('Duplicate rows:', rep['n_duplicate_rows'])
df['Class'].value_counts().plot.bar(log=True, title='Class counts (log scale)')
plt.show()
```

## Amount and Time by class

`Time` drives the temporal split and stays out of the model features by default (config.py explains why). `Amount` feeds the cost-aware threshold.

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
df.loc[df.Class == 0, 'Amount'].plot.hist(bins=80, ax=ax[0], alpha=.6, label='legit', log=True)
df.loc[df.Class == 1, 'Amount'].plot.hist(bins=80, ax=ax[0], alpha=.6, label='fraud', log=True)
ax[0].set_title('Amount (log count)'); ax[0].set_xlabel('Amount'); ax[0].legend()
df.loc[df.Class == 0, 'Time'].plot.hist(bins=80, ax=ax[1], alpha=.6, label='legit')
df.loc[df.Class == 1, 'Time'].plot.hist(bins=80, ax=ax[1], alpha=.6, label='fraud')
ax[1].set_title('Time (s elapsed)'); ax[1].set_xlabel('Time'); ax[1].legend()
plt.tight_layout(); plt.show()
```
